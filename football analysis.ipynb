Kaggle Dataset: International Football Results

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results.csv")
df.head()
```

## ⚽ Goals Analysis

```
df["total_goals"] = df["home_score"] + df["away_score"]
```

## 🏆 Match Results

```
def match_result(row):
    if row["home_score"] > row["away_score"]:
        return "Home Win"
    elif row["home_score"] < row["away_score"]:
        return "Away Win"
    else:
        return "Draw"

df["result"] = df.apply(match_result, axis=1)
```

```
(df["result"] == "Home Win").mean() * 100
```

```
df["result"].value_counts(normalize=True) * 100
```

```
home_wins = df[df["home_score"] > df["away_score"]]["home_team"]
away_wins = df[df["away_score"] > df["home_score"]]["away_team"]

all_wins = pd.concat([home_wins, away_wins])
all_wins.value_counts().head(1)
```

## 📊 Visualizations

```
df["total_goals"].hist(bins=15)
plt.title("Distribution of Goals Per Match")
plt.xlabel("Total Goals")
plt.ylabel("Frequency")
plt.show()
```

```
df["result"].value_counts().plot(kind="bar")
plt.title("Match Outcomes")
plt.xlabel("Result")
plt.ylabel("Number of Matches")
plt.show()
```

```
all_wins.value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Teams by Total Wins")
plt.xlabel("Team")
plt.ylabel("Wins")
plt.show()
```
